# **Hard to Understand? Let's Use a Real-World Example!**

Think of a function like a recipe in a restaurant. Let's break it down:

1. Imagine you are in a restaurant, and every time a customer orders a dish, the chef follows a **fixed recipe** to prepare it.

2. Instead of writing down the steps every time in order to order, the chef just **refers to the recipe**, making the process **faster and more efficient.**

3. In Python, a f**unction works just like a recipe** – it is a **set of instructions** that performs a specific task whenever needed.

4. Once a function (recipe) is defined, you can **reuse it multiple times without repeating the steps**, just like the chef making the same dish again and again.

5. If you ever need to make small changes to a dish, you can **modify the recipe** instead of rewriting the whole thing – this is like **updating a function** to improve or extend its behavior.

6. Functions **save time, reduce errors, and keep the code organized**, just like recipes make cooking more efficient in a busy restaurant!


In programming, functions help us write clean, efficient, and reusable code. Just like a chef doesn't need to remember every dish's steps from scratch, you just need to rewrite every time – we just call the function!

Why Use Functions?

Just like in a restaurant, where recipes help chefs work **faster and more efficiently**, functions help programmers write **cleaner and reusable code.**

✅ **Benefits of Functions (Using Our Restaurant Analogy)**

**Avoids Repetition** – Instead of writing the same steps repeatedly, we create a function (recipe) once and reuse it.

**Modular & Organized**– Just like restaurants have different recipes for different dishes, programs have functions for different tasks.

**Improves Readability & Debugging** – If there's an issue with a dish, the chef fixes the recipe instead of checking every single order. Similarly, debugging a function is easier.

**Code Reusability** – A restaurant doesn't rewrite the recipe for every new customer; it just follows the same process. Likewise, we can reuse functions in different parts of our program.

**Enhances Collaboration** – Different chefs can prepare different dishes simultaneously, just as developers can work on different functions.

# **3.any()**
- **any()** function returns True if at least one element of an iterable is true. 

```python
brush= input("Did you brush your teeth? (yes/no): ")
shower= input("Did you take a shower? (yes/no): ")
reading= input("Did you read a book? (yes/no): ")
# Convert to boolean
brush_done = brush.strip().lower()=="yes"
shower_done = shower.strip().lower()=="yes"
reading_done = reading.strip().lower()=="yes"
# Use any() to check if at least one task is done
if any([brush_done, shower_done, reading_done]):
    print("You have completed at least one of your daily tasks.")
elif all([brush_done, shower_done, reading_done]):
    print("You have completed all your daily tasks.")
# Use any() to check if at least one task is done
else:
    print("Keep working! You haven't completed any of your daily tasks yet.")
```

# **14.map ()**
* Use map() when:

* You want to transform all items in a list (or tuple)

* You want to avoid writing loops

* You are applying the same function to everything

## 1. Double All Numbers


```python
# Multiply each number by 2
numbers = [1, 2, 3, 4, 5]
doubled_list= list(map(lambda x: x * 2, numbers)) # Using map() function to double each number in the list
print(f"Doubled List: {doubled_list}")  # Print the doubled list
```

## 2. Convert Names to Uppercase

```python
names= ["Alice", "", "Bob", "Charlie", "", "David"]
upper_names = list(map(lambda x : x.upper() , names))
print(f"Uppercase Names: {upper_names}")  # Convert names to uppercase and print them
```

## 3. Convert String to Integer

```python
num_strings = ["1", "2", "3", "4", "5"]
#int_numbers = list(map(lambda x: int(x),num_strings))
int_numbers = list(map(int ,num_strings)) # Other way to convert string numbers to integers

print(f"Converted Integers: {int_numbers}")  # Convert string numbers to integers and print them
```

## 4.Calculate squares of numbers

```python
nums = [3, 5,7,9]
squared_nums = list(map(lambda x : x**2 , nums)) # Calculate squares of numbers using map
print(squared_nums) # Output: [9, 25, 49, 81]
```

## 5. Add Two lists Elements-wise

```python
list1 = [1, 2, 3]
list2 = [4,5,6]
sums = list(map(lambda x , y : x + y,list1, list2))
print(sums)
```

# **17.memoryview()**
 When to use memoryview()?
When working with binary data (files, images, bytes)

When you want to read or modify data without copying it (faster)

Used in advanced tasks like image processing, networking, etc.

```python
data = bytearray("hello", "utf-8")

view = memoryview(data)

print(view[0])        # ➜ 104 (ASCII code for 'h')
print(view[1])        # ➜ 101 (ASCII for 'e')
```
