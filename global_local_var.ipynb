# **1. Global and Local Variables**

## Global Variable:
* Defined outside of all functions.

* Can be used anywhere in the code, including inside functions (with some rules).

```python
# Global and Local Variables in Python
x = 10 # Global variable
def my_function():
    y  = 5 # Local variable
    print(f"Inside function: x = {x} , y = {y}")
my_function()
print(f"Outside function: x = {x}")
#print(f"Outside function: y = {y}")  # This will raise an error since y is not defined outside the function
```

## Local Variable:
* Defined inside a function.

* Only accessible within that function.

* Not known outside the function.

```python
# Local Variable:
x = 10 # Global variable
def my_function():
    y  = 5 # Local variable
    print(f"Inside function: x = {x} , y = {y}")
my_function()
#print(f"Outside function: y = {y}")  # This will raise an error since y is not defined outside the function
```

## **2. How Python Handles Variables Scope?**
* Python’s LEGB Rule
* Python looks for a variable in the following order:
* L → Local  
* E → Enclosing  
* G → Global  
* B → Built-in


 ## 1. Local Scope (L)
🔹 Variable defined inside a function — used only inside that function.
* Think of it like: A notebook in a student’s desk – only they can see it.     

```python
# Local Variable:
def my_func():
    local = 5 # Local Variable
    print(f"Inside function: x = {local}")
my_func()
# print(local) # It will raise an error since local is not defined outside the function

```

## 2. Enclosing Scope (E)
* A variable defined in an outer function, and used in an inner function.
* Think of it like: A classroom drawer – all students inside the class can use it.

```python
# Enclosing scope example
def outer():
    y = 20 # Enclosing variable
    print(f"Outer function: y = {y}")
    def inner():
        print(f"Inner function: y = {y}")
    inner()
outer()
```

## 3. Global Scope (G)
* A variable defined outside all functions – can be used anywhere.
* Think of it like: A school notice board – everyone in school can read it.

```python
z = 20  # Global variable

def display():
    print(z)

display()
print(z)
```

## 4. Built-in Scope (B)
* Predefined functions and keywords provided by Python (like print(), len(), max(), etc.)

```python
print(len("Hello"))  # len() is a built-in function
```

# **3. Modifying the global Variables inside Function**
* To modify a **Global Variable inside a function** we must use the **`global`** 

## Modify using global Example:


```python
# Modifying the global Variables inside Function
# To modify a Global Variable inside a function we must use the global keyword
counter= 0 # Global Variable
def increment():
    global counter # Declare counter as global to modify it
    counter += 1
    print(f"Counter inside function: {counter}")
increment()
print(f"Updated Counter: {counter}")  # Output: 1
```

# **4. Best Practices for Using Global and Local variables**
&#9745;

✔️ Prefer **Local Variables** whenever possible to **avoid unintended side effects.**

✔️ Minimize the use of `global` variables to keep code modular and maintainable.

✔️ Use **function arguments** to pass values instead of relying on lobal state.



# **5. Understandinfg** `is` (Identity comparison)
* **The `is` operator checking Object identity**

```python
# is opertor checking Object identity
a = [1, 2, 3]
b = a # both point to the same values
c = [1, 2, 3]
print(a is b) # Output: True
print(a is c) # Output: False
```

# **6. Understanding `==` (Value Comparison)**
Example: The `==` Checking Values Equality 

```python
# == operator
a= [1, 2, 3]
b = [1, 2, 3]
c = [1,2,3]
a = c
print( a == b)# oupu: True
print(a == c) # Output: True
print(c is a)
print(a is b) # Output: False (different objects in memory)
```

# **7. Real world scenarios Where `is` and `==` Matter**
scenario 1: Comparing Immutable Objects (`is` works for small integers and strings)


```python
a = 1000
b = 1000
print(a is b ) # Output: False (different memory locations)

a = 255
b = 255
print(a is b) # Output: True (same memory location for small integers)
# Reason: Python caches small integers (typically -5 to 256) 
c = 10 
d = 10
print(c is d) # Output: True (same memory location for small integers)
```
