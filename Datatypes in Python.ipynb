## 5.4 Set
### Definition & Charactristicks
1. A Set is created using curly braces {}  or set() constructer.
2. It is unordered and does not allow duplicate values.
3. It is mutable: You can add or remove elements from a set.
### creating a basic operations


```python
# Creating a set 
empty_set = set()
number_set = {1,2,3,2} # Duplicates remove automatically 
mixed_set = {2, "two", 3.0, True} # you can store immutable types like tuples
print(empty_set) # Output: set()
print(number_set) # Output: {1, 2, 3}
print(mixed_set) # Output: {1, 3.0, 'two'}
```

### Adding Removing Elements


```python
number_set.add(4) # Adding 4 to the set
#numbber_set.remove(3) # Removing 2 from the set rasing error if not exists
number_set.discard(3) # Removing 3 from the set not raising error if not exists

print ('Updated set:', number_set)
```

### Set Operations
* Sets supports the set operations like union , intersection, difference etc

```python
set_a = {1, 2, 3}
set_b = {3, 4, 5}
# Union all elements from both sets
print("Union:", set_a.union(set_b)) or print("Union:", set_a | set_b)
# Intersection common elements in both sets
print("Intersection:", set_a.intersection(set_b)) or print("Intersection:", set_a & set_b)
# Difference elements in set_a but not in set_b
print("Difference:", set_a.difference(set_b)) or print("Difference:", set_a - set_b)
# Symmetric Difference elements in either set_a or set_b but not both
print("Symmetric Difference:", set_a.symmetric_difference(set_b)) or print("Symmetric Difference:", set_a ^ set_b)
```

### Best Practices
1. Use sets when you need to efficiently test for membership (in check) and remove duplicates.
2. Avoid storing mutable items in a set---items must be hashable (immutable).
3. Use set operation for cleaner and more efficient code when dealing with memberships and overlape.

## Using `Enumerate()` in For Loop
Some times you need both the item and index. Python built in `enumerates()` functions makes its easy.

```python
# Example 2 : Using enumerates to get both index and value from a list.
fruits = ['apple', 'banana', 'cherry']
for index, fruit in enumerate(fruits):
    print(f'Index {index}: {fruit}')
```

```python
# List of Fruits names
fruits = ['apple', 'banana', 'cherry']
# Initialize a counter value
index = 0

# Iterate over each fruits in the list
for fruit in fruits:
    print(f"Index {index}: {fruit}") # Print the current index and fruit
    index += 1  # Manually increment the index counter
```

## Iterating over Dicionaries
Dictionaries are keys value pairs. You can iterate over.
* Keys: By default, looping over a dictionary return its keys.
* Values: Using `.Values()`
* Key-Value Pairs: Using `.items()`

```python
# Iterating over a dictionary
person = {
    "name" : 'Alice',
    "age" : 25,
    "city" : "New York"
}
# Iterating over keys (default behavior)
print("Iterating over keys:")
for key in person:
    print(f"{key}: {person[key]}") # Accessing the value using the key
    
```

```python
# Iteratig over keys-value pairs using .items() method
for key, value in person.items():
    print(f"{key}: {value}") # Accessing both key and value
```

# 3. While loops
* While loop overview:
The `while ` loop  runs as long as a specified condition is `True`.

```python
# Example 4: a basic while loop that counts from 0 to 4
count= 0
while count < 5:  # Continue until count is less than 5
    print('count is:', count)# Print the current count
    count +=1 # # Increment the count by 1
```

## control statement
1. `breaks`:
   * Exits the loop immediately
   * Usefull when a certain conditions is met and further processing is unnecessary.
2. `Continue`:
   * Skips the current iteration and moves to the next one.
   * Ideal for bypassing specific condition without stopping entire loop.
3. `else`:
   * Runs when the loop complete naturally (i.e., it is not terminated by a `breaks` statement).
   * Can be used to execute code after the loop finishes processing.

```python
# Example 5: Using break, continue, and else in a loop
for i in range(10):
    if i ==3:
        print("Skipping 3")
        continue
    if i==8:
        print("Breaking at 8")
        break
    
else:
        # This block executes only if the loop was't terminated by break
    print("Loop completed without break.")
```
