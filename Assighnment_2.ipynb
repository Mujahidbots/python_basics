Question 2: Operators
Write a Python program that:
1. Accepts two numbers as input from the user.
2. Performs and prints the result of all arithmetic operations (addition, subtraction, multiplication, division,
modulus, floor division) between these two numbers.
3. Uses comparison operators to check if the first number is greater than the second, and if they are equal.
4. Uses logical operators to combine two conditions (e.g., the first number is greater than the second, and the
second number is less than 10).

```python
first_num=int(input("Enter a first number:"))
second_num=int(input("Enter a second number:"))
# Perform Common operations
# Addition
addition= first_num + second_num
# Subtraction
subtract= first_num - second_num
# Multiplication
multiply= first_num * second_num
# Division
division= first_num  / second_num
# Modulas
modulas=first_num % second_num
# Flow division
flow_division= first_num // second_num
# Print results of operations
print(f'The addition of two numbers is: {addition}')
print(f'The subtraction of two numbers is:{subtract}')
print(f'The multiplication of two numbers is {multiply}')
print(f'The Modulas of two numbers is {modulas}')
print( f"The division of two numbers is: {division}")
print(f"The Flow division of two numbers is: {flow_division}")
# Logical operations
num1=first_num > second_num
num2=first_num == second_num
print(f"First number is greater than Second: {num1}")
print(f"First number is equal to second:{num2}")
print(f"Both conditions are true: {first_num > second_num and second_num < 10}")
```

Question 3: Loops
Write a Python program that:
1. Accepts a list of integers from the user.
2. Loops through the list and prints out each number.
3. If a number is greater than 10, skips it using the continue statement.
4. Stops the loop if the number is 20 using the break statement.
5. After the loop ends, prints a message that the loop ended naturally

```python
# Take a list of numbers from user
user_list=input("Enter a list of numbers seperated by spaces:")
# Map the input string to integers list
numbers= list(map(int,user_list.split()))
# Iterate using for loop and print numbers less than 10
for i in numbers:
  if i < 10:
    print(i)
# print numbers greater than 10
  if i > 10:
#  skip numbers greater than 10 using continue
    print(f"Skipping {i}")
    continue
# break the loop if number is 20
  if i ==20:
    break
# print the result
print("Breaking at 20")
print("loop endns Naturally")

```
