# **Loops in python**

## 1. Nested loop

```python
# use nested loop
for i in range(1,6):
  for j in range(0,i):
    print("*", end=" ")
  print("")
```

## 2. For loop and break

```python
# iterate through each
for i in range(1,11):
  if i ==5: # if i is 5 it breaks
    break
  print(i)
```

```python
# using continue
# itrate through each
for i in range(1, 11):
  if i==5: # if i is 5 it continue
    continue
  print(i) # print all numbers skiping 5
```

```python
for i in range(1,11): # Using for loop to pass if we not have any input
  pass
```

## 3. All number skip muliple of 3

```python
# itrate through 1 to 50 using for loop
for num in range(1,51):
  #check conditions
  if num % 3 ==0: # skipping the multiply of 3 continue statement
    continue
  print(num)
```

## 5. Control Flow loop

```python
# To get User input negative
user= int(input("enter a number:"))
# if input is positive
while user > 0:
  if user ==0: # skip user input=0
    continue
  user= int(input("Enter a number:"))

print("Ok This is a negative number.") # Getting negative number
```

### Control flow

```python
# Use While loop and control flow
# continiue, break
while True:
  try:
    uinput= int(input("Enter positive a number: "))
  except ValueError:

    print("Enter only number not Alphabets.") # Accept Alphabets
    continue # skip Alphabet
  if uinput ==0:
    continue # skip uinput= 0
  elif uinput > 0:
    print(f"{uinput}  is positive number") # Only accept positive
  else:
    print(f"{uinput} is negative number! and Finish!")
    break
```

## 6. Rolling a Dice

```python
# Import random module
import random
while True:
  roll = random.randint(1,6) # Roll a dice and give a random number
  # Making condition
  if roll ==6:
    print(f'You rolled a {roll}.Game ends!')
    break   # Break loop at 6
  if roll ==1:
    continue # Skip showing 1

  print(f"You rolled a {roll}")
```

## 7. Calculations

```python
while True:
  # Get calculation input from user
  userinputs=input("Enter a calculation with +,-,/, (Type 'exit' to stop) ")
  if userinputs.lower()=="exit": # Exit at type
    break
  try:
    result=eval(userinputs) # use eval function to perform calculation
    print(f'Result: {result}') #
  except:
    print("Invalid input") # Except an error
    continue
```

## 8. Prime number

```python
# Itrate through each number
for i in range (2,101):
  for j in range (2,i): # use nested loop
    if i % j ==0:
      break  # break if not a prime number
  else:
    print(i)
```

### Prime number using Exception

```python
try:
  user=int(input("Enter a number: ")) # Take input number from user
  for i in range(2,user): # # Itrate through each number
    if user % i==0:
      print(f"{user} is not a prime number") # If not a prime number
      break
    else:
      print(f"{user} is Prime number") # prime number
      break # Break loop
except:
  print("Invalid input. Please Enter a valid input") # Except ValueError
```
