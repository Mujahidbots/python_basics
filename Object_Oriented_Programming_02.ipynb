# **`Self` = Refrense to the current object**

`self` is conventional name for first argument of every instance method in python. It allways point to the particular Object on which the method is called, giving you direct access to that object's attributes and other methods.

```python
# Instance Method
class House:
  def __init__(self, colour):
    self.colour=colour
  # Use Self to create method
  def show_colour(self):
    print(f"The House is painted with {self.colour}")
# Create House Instance
my_house=House("Brown")
my_house.show_colour()
```

# **INstance Method = Methods Bound to an Instance**

Instance methods are functions defined inside a class that operate on an instance of the class they take `self` as their parameter, which refers to the specific intance calling the method. These methods can access and modify the atributes of the instance and call other insance methods.

```python
class House:
  # constructer to initialize the house with a specific color
  def __init__(self, color):
    self.color=color # Assign a color to the house instance
  # Instance method to display the color of the house
  def show_color(self):
    print(f"The house is painted with {self.color}") # Access and print house
# Create an instnce of the house with the color "Red"
my_house = House("red")
my_house.show_color()
```

#**Class Methods = Bound to the Class, Not the Object**

A class method is a method that is bound to the class itself, not any one instance of the class. It can access or modify the class-level state, which is shared among all instances. Unlike instance methods that receive self, class methods receive cls as the first argument—referring to the class, not an individual object.

To define a class method, you use the @classmethod decorator above the method definition. Class methods are commonly used for factory methods or when you want behavior that affects the class as a whole rather than just one instance.

```python
class House:
  def __init__(self, color , age):
    self.age=age # Instance attributes
    self.color=color # Instance attributes
  @classmethod
  def from_construction_year(cls, color , construction_year):
    # This class method creates a House object by calculating its age
    return cls(color , 2025 - construction_year)
# Creates a new House using class methosd as factory
my_house = House.from_construction_year('White', 2015)
print(my_house.age)
```

# **Static Method = Independent Utility Tools**

A **static method** is method that belongs to class but does not access or modify the class or any of its instances. It does not take `self` (the instances ) (the class) as its first argument. Instead, static method behaves like functions, but they are logically grouped with class because they are perform tasks relevent to class's purpose.

Static method are ideal for utility operations__ validations , calculation , or class's internal state. You define them using the `@staticmethod` decorated.


```python
# Static method
class House:
  def __init__ (self, color, area):
    self.color= color
    self.area=area

  @staticmethod
  def is_valid_area(area):
    return area > 0
# Use the method to check if a given area is valid before creating House
print(House.is_valid_area(120))
print(House.is_valid_area (-50))
```

```python
class Rectangle:
  def __init__(self, width , height):
    self.width=width
    self.height=height
  @staticmethod
  def is_valid_dimension(dimension):
    return dimension > 0
  @classmethod
  def create_rectangle(cls, width , height):
    if Rectangle.is_valid_dimension(width) and Rectangle.is_valid_dimension(height):
      return cls (width , height)
    else:
      raise ValueError ('width and height must be positive')
rect = Rectangle.create_rectangle(3, 6)
#print(rect)
print(rect.width, rect.height)
```

### Comparison Table

| Feature               | Instance Method                      | Class Method                                       | Static Method                          |
| --------------------- | ------------------------------------ | -------------------------------------------------- | -------------------------------------- |
| **Binding**           | Instance of the class                | Class itself                                       | Neither instance nor class             |
| **Required Argument** | `self`                               | `cls`                                              | None                                   |
| **Use Cases**         | Access/modify instance-specific data | Access/modify class-specific data, factory methods | Utility functions related to the class |
