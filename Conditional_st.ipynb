# 1. Traffic signals

```python
# Traffic signals example
# Define a traffic light colour
traffic_light= "Green"
# Check conditios
if traffic_light=="Green":
  print ("You can go!") # Cars can Go!
else:
  print("You must stop!") # Cars must Stop!
```

## Full Traffic signals code

```python
# Pedestrian signal example
# Define a pedestrian signal
pedestrian_signal= "Flashing"
# Check the conditions
if pedestrian_signal=="Walk" and pedestrian_signal !="Flashing": # Pedestrian go walk , when no Flashing
  print("Pedestrians can cross safely! ")
# If pedestrian signal is flashing, they should not cross
else:
  print("Pesestrian should WAIT!")

```

### Final code for traffic signals

```python
# Traffic signals example
# Define a traffic light colour
traffic_light="Green" # If traffic  light is Green, cars can go
# Check conditions
if traffic_light == "Green":
  vehical_type="Ambulance" 
# If traffic light is Green and vehical type is Ambulance, then they can go
# Check vehical type AND traffic light
if traffic_light== "Green" and vehical_type=="Ambulance":
  print("You can Go!")
else:
  print("You must stop!") # If traffic light is not Green or vehical type is not Ambulance, then they must stop
```
