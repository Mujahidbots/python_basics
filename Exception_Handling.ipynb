* **1.Syntax Error** (Parsing Error)

  * These are detected **before excecution.**
  * Example: A missing closing paranthesis, incorrect indentation, or incorrect keyword usage can trigger a `SyntaxError`
* **2.Runtime Error** (Exception)
  * These occurs **while the program is running**
  * The syntax is correct but something unexpected happens during excecution, such as division by zero or accessing undefined variable.

# **4. Basic `try-except` structure**
**How does `try-except` Work?**

The `try` block contains code that might raise an exception, while the except block handles the exception if it occurs. It prevent the program from stopping unexpectadly.

**Example:** is a simple demonstration using `try-axcept` to handle errors.

```python
try:
  # Prompt the user to enter a number
  num = int(input("Enter a number: "))
  # Attempt to devide 10 by the entered number
  print(10 / num)
except ZeroDivisionError:
  # Handle the case where the user enters zero
  print("Error! You  cannot divide by zero.")
```

# **5. Handling Multiple Exceptions**
When different types of errror can occur, we can handle each seperately multiple `except` blocks.

**Example:** The following code demonstrates how to handle both `ZeroDivisionError` and `ValueError` saperately.

```python
try:
  # Prompt the user to enter the the first number
  num1= int(input("Enter a first number: "))
  # Prompt the user to enter the the second number
  num2 = int(input("Enter a second number: "))
  # Attempt to devide the first number by the second number
  result = num1 / num2
  print("Result:", result)
except ZeroDivisionError:
  # Handle the case where the second number is zero
  print("Error: Cannot divide by zero!")
except ValueError:
   # Handle the case where the input is not a valid integer
  print("Error: Invalid input! Please Enter  number only.")
```

# **6. Handling Any Exception**
**Catching Any Exception**
Sometimes, we don't know which error might occur. We can catch **any exception** using `except exception as e`. H catching all exceptions should be used cautiously, as it can hide important bugs.

**Example:** Below is an example of handling any exception.

```python
try:
  num= int(input("Enter a number: "))
  result = 10 / num # May raise ZeroDivisionError
  print("Result: ", result)
except Exception as e:
  print("An Error occurred: " , e)
```

# **7. Using the `Finally Block`**
The finally code block is also a part of exception handling. When we handle exception using try and except block, we include a finally block at the end. The finally block is allways executed , so it is generally used for doing the concluding ta closing file resources or closing database connection or may be ending the program excecution with a delightfull massage.

```python
try:
  # Attempt to open the file in read mode
  file = open ("example.txt" , "r")
  content= file.read()
except FileNotFoundError:
  # Handle the case where the file does not exist
  print("File not found!")
finally:
  # This block runs no matter what
  print("Closing the file...")
  file.close()
```

# **8. Raising Custum Exceptions**
**Creating Custum Exception**
We can define our own exception using the `raise` keyword when specific condition are not met.

**Example:** The following functio raise a `ValueError` if a negative amount entered.

```python
def withdraw (amount):
  # Check if the amount is negative
  if amount < 0:
    # Raise a ValueError if the amount is negative
    raise ValueError("Amount cannot be nagative!")
  # Print the withdrawal amount
  print(f"Withdrawing ${amount}")
try:
  # Attempt to withdraw negative amount
  withdraw (100)
except ValueError as e:
  # Handle the Value Error and print the Error massage
  print(f"Error: {e}")
```

## 6. Contact Book
Description: Store names and phone numbers in a dictionary.

```python
contacts = {}

try:
    while True:
        print("--- Contact Book ---")
        print("1. Add Contact")
        print("2. Search Contact")
        print("3. View All Contacts")
        print("4. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            name = input("Enter name: ")
            phone = input("Enter phone number: ")

            # Validate phone number
            if not phone.isdigit() or len(phone) < 7:
                raise ValueError("Invalid phone number! Must be digits and at least 7 characters.")

            contacts[name] = phone
            print("Contact added successfully.")

        elif choice == "2":
            search_name = input("Enter name to search: ")
            try:
                print(f"{search_name}: {contacts[search_name]}")
            except KeyError:
                print("Error: Contact not found.")

        elif choice == "3":
            if contacts:
                print("\n--- =ll Contacts ---")
                for name, phone in contacts.items():
                    print(f"{name}: {phone}")
            else:
                print("No contacts saved yet.")

        elif choice == "4":
            print("Exiting Contact Book...")
            break

        else:
            print("Invalid choice. Please select 1-4.")

except ValueError as ve:
    print("Error:", ve)

except Exception as e:
    print("An unexpected error occurred:", e)
```
