# Comparison Operators
The output of comparison operators is a boolean value (True or False)

```python
# Comparison operators
x , y = 5, 10
print("x > y:", x > y)
print("x < y:", x < y)
print("x == y:", x == y) #Double equal sighn used for comparison if two values are equal
print("x != y:", x != y)
print("x >= y:", x >= y)
print("x <= y:", x <= y)
```

## Identity Operators
* is and is not operators

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list1 is list2 # Check if list1 is the same as list2
x = 100
y = 101
x is not y # Check if x is not the same as y
print("x is not y:", x is not y) # Identity operator to check if x
```
