# **What are Decorators?**
* Decorators in python are functions that **modify the behavior** of other functions **without changing their code.**They alow us to enhance or extend the function in reusable way.

**Key Features**
* Allow code **reusability.**
* Improve **Code readabilty**
* Used for **logging authentication, timing function and more.

# **2. Why do we use decorators?**
Decorators help us **avoid code duplication** by adding extra functionality without modifying existing code.

**Use cases of Decorators:**
* **Measuring execution time of a function**
* **Passed as an arguments**
* **Checking a user authentication before executing a function**

**Example: A simple decoraor that logs function calls**




```python
# Define a decorators function that takes another function as an argument
def log_decorator(func):
  # Define a wrapper function that will modify behavior of the orignal function
  def wrapper ():
    print(f"Calling Function: {func.__name__}") # Print the name of function being calling
    return func() # Call the original function
  return wrapper # Return the modified function

  # APPly the decorator to "say_hello" function using @logdecorator
@log_decorator
def say_hello():
  print("Hello!") # This function simply print hello!

# call the decoratored function
say_hello()


```

## 5. Using Multiple Decorators
Multiple decorators can be applied **by stacking them** on top of a function.

**Example: Using multiple decorators to log and time function execution**




```python
import time
```

```python
# Using decorators for time
#import time
# Decorators to log patient details before consaltation
def log_decorator(func):
  def wrapper (patient_name):
    print(f"Recording patient details: {patient_name}")
    return func(patient_name)
  return wrapper
# Decorators to measure consultation time
def timer_decorator(func):
  def wrapper(patient_name):
    start = time.time() # Record start time
    result = func(patient_name) # Call the orignal function
    end = time.time() # Record end time
    print(f"Consultation time: {end - start:.4f} seconds")
    return result
  return wrapper

# applying multiple decorators to a doctor's consultation
@log_decorator # First, log patient details
@timer_decorator # Then, measure consultation time
def doctor_consultation(patient_name):
  print(f"Doctor is consulting {patient_name}...")
  time.sleep(1) # simulating consultation time
  print(f"Prescription given to {patient_name}")

  # Simulating a patient visiting the doctor
doctor_consultation("Alice")

```

## ✅ 1. Access Control (Admin Check)
Goal: Only allow the function to run if user is "admin".

```python
# Decorators for admin check
def admin_check (func): # Define function
  def wrapper (user):   # Make a wrapper
    if user =="admin":
      print("Access allowed!")
      return func(user)
    else:
      print("Access denied! ")
  return wrapper




```

```python
@admin_check
def view_dashboard(user):
  print( f"Welcome! to the dashboard, {user}")
view_dashboard("admia")
```

## ✅ 2. Repeat Function

```python
# Outer function that takes a parameter  `times ` (how many times to repeat)
def repeat (times):
  def decorator (func): # define decorator
    def wrapper (*args , **kwargs):
      for __ in range (times):
        func(*args , **kwargs)
    return wrapper
  return decorator
# Applying function
@repeat(12)
def greet():
  print("Hello")
# Calling function
greet()
```

## ✅ 4. Only Even Allowed
Goal: Only run the function if the number given is even.

```python
# Decorators that print only even numbers
def only_even_decorator(func):
  def wrapper (numbers):
    if numbers % 2 ==0:
      return func(numbers)
    else:
      print("Only Even numbers is allowed: ")
  return wrapper
@only_even_decorator
def square_numbers(number):
  print (f"Square of {number} is {number * 2}")
# Testing function
square_numbers(1)
```

```python
# function decorator to Save function results for repeated inputs
def catch_result(func):
  catch={}    # Create empty dictionary to store previous results
  def wrapper(n):
    if n in catch:
      print(f"Fatching from catch {n}")
      return catch[n]  # Return saved result
    else:
      print(f"Calculating: {n}")
      result=func(n)  # Calculate new result
      catch[n] = result # Save in cache
      return result
  return wrapper
@catch_result
def square(a):
  print (a * a)
square(2) # Calculates
square (2) # Uses catch

```

## ✅ 7. Count Function Calls
Goal: Count how many times a function is called.

```python
# Count how many times function is called
def count_calls(func):
  count = 0
  def wrapper ():
    nonlocal count
    count += 1
    print (f"Function called {count} times!")
    return func ()
  return wrapper
# Applying decorators
@count_calls
def greet():
  print("Hello!")
greet()
greet() # Function called 2 times!

```

## ✅ 8. Delay Function
Goal: Add a delay before running the function.

```python
import time
```

```python
# Function using decorator for delay
def delay_function(func):
  def wrapper():
    print("Please Wait...")
    time.sleep(2)
    return func ()
  return wrapper
@delay_function
def greet():
  print("Hello World! ")
greet()


```
