# **Object Oriented Programming**
For example:

A blueprint specifies that every house will have a door, windows, and rooms.

However, the blueprint itself is not a house—it is just a plan.


Similarly, in Python, a class defines the structure and behavior (methods) that its objects will have, but it does not represent an actual object until you create one.

Think of the class as the template for creating objects. Each object created from the class can have its own unique property while following the same structure defined by the class.


```python
# Syntax
class House:
  pass
```

## **Object = Instance of a class**
An **Object** is specific instance of a class. When you create an object, you are essentially building something based on the blueprint (class)

For Example:
  
  * If the class is blueprint for a house built from that blueprint.
  * Each object can have its own unique properties (e.g, Color , size ) while the structure define by the class.
In python , you create an object by calling the class as if it were a function. This process is called **instantiation.**


```python
my_house = House()
```

## **Attributes = Object's Data**

Attributes are the data stored inside an object. They represent the state or properties of the object. They represent the state of the properties of the object.
For Example:

   * If object is car , attributes could be its `color` . `brand` or `year`.

   * Each object can have different values for these attributes.

In python, attributes are typically defined in the `___init___` method of class.

```python
class House:
  # The constructer (__init___) method initializes the attributes of the object
  def __init__(self, color , size):
    self.color = color  # Attributes to store the color of the house
    self.size = size    # Attributes to store the color of the house
my_house = House ("blue" , "large")
# print the color of the house object
print(my_house.color)
```

## **Methods = Objects Behavior**
Methods defines the behavior of an object. They are the functions defined inside the class and are used to perform actions using objects data.
For Example:
  
  * If the object is car the methods coud be `drive()` and `stop()`.
  * Mehods allow objects to intract with their data and perform specific task.
  
In python, methods are defined like regular functions but allways take `self` as their first parameter, which represent the instance of a class.

```python
class House:
  # The constructer (__init__) method initializes the attributes of the objects
  def __init__(self, color , size):
    self.color=color
    self.size = size
  # Methods to describe the house
  def describe(self):
    print(f"The house is {self.color} and {self.size}")
# Create an object of the house class with specific color and size
my_house = House("Blue", "Large")
print(my_house.color)
my_house.describe()

```
