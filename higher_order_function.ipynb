# **1. What are Higher-Order Functions?**
* In python, Function that **take another function as an argument** or **return a function as output** are called Higher-Order **function**.These help make code more readable and efficient.

**Commen Higher-Order Functions:**

* `map():` Applies a function to every item in an iterable (like a list or tuple).
* `filter():` Filters elements based on a condition.
* `reduce():` Reduce a sequence to a single value.
* `zip():` Combines mutable iterables into tuples.
* `enumerates:` Adds an index to an iterable.


# **2. Understanding `map()` Function**
*  When to use:

👉 When you want to change or transform each item in a list.

The map() function applies a given function to **each item** in an iterable (like a list, tuple, or string) and returns a new iterable with the transformed values.

**Syntax:**

map(function , iterable )

**Example: Applying `map()` to doube each number in a list.**

```python
# Function to a double number 
def double(x):
    return x * 2
# Using map() to double number in a list
numbers = [1, 2, 3, 4, 5]
doubled = list(map(double , numbers))
print(doubled) # Output: [2, 4, 6, 8, 10]

```

# **3. Understanding `filter()` Function**
**Syntax:**

* When to use:

When you want to select/filter items from a list based on True/False condition.

filter (function, iterable)

**Example: Filtering even numbers from a list**


```python
# Function to check if numbeer is even
def is_even(x):
    return x % 2 == 0
# Using filter() to get even numbers from a list
numbers = [1, 2, 3, 4, 5]
even_numbers = list(filter(is_even, numbers))
print(even_numbers) # Output: [2, 4]
```

# **4. Understanding `reduce()` Function**
The `reduce` function **reduce** an iterable into a **String value** by applying a function

When to use:
* When you want to combine all items in a list into one result (sum, product, etc.)

* You must import it from functools.

**Syntax:**
* reduce(function , iterable)
  
**Example: Finding the product of all numbers in a list.**

```python
# From functools import reduce
from functools import reduce
def multiply(x, y):
    return x * y
# Using reduce() to multiply all numbers in a list
products= [1,2,3,5,6]
products= (reduce(multiply , products))
print(products) # Output: 180
```

# **5. Understanding `zip()`** function
* When to use:
* When you want to pair items from two or more lists.

```python
# Using zip() to pair names and ages
names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30, 35]
paired = list(zip(names , ages)) # Output: [('Alice', 25), ('Bob', 30), ('Charlie', 35)]
print(paired)
```

# **6. `enumerates` Function**

* The `enumerate()` function adds an index to an iterable, making it easer to track element position.
  
 **Syntax:**

 enemurate(iterable, start = 0)

 **Example: Using `enumerate()` to get index and value from list**


```python
# Using enumerates() to print index and value
fruits = ["Apple" , "Banana", "cherry"]
for index, fruit in enumerate(fruits, start=1):
    print(f"{index} : {fruit}")
```

## 3. Sort Names by Length (sorted)

```python
# Function to sort words by length in descending order
def sorted_by_length(words):
    return sorted(words, key = len, reverse=True)
sorted_by_length(["apple", "banana", "kiwi", "cherry", "blueberry"])
```
