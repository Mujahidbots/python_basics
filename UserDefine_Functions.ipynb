# **Functions**
## **User Define Function**
Userdefine functions

There are three types of functions.
1. Default Functions
2. Keyword Functions
3. Required functions
4. Variable Length functions


## **1. Default Functions**
If a function parameter has a default value, you don’t have to pass it when calling the function.

If you don’t provide a value, the default is used.

## **2. Keyword Arguments**
Normally, arguments are passed in order (positional).

But with keyword arguments, you explicitly mention the parameter name while passing the value.

This makes the code more readable and avoids confusion.

## **3. Required Arguments (Positional Arguments)**
These are mandatory arguments that must be passed when calling the function.

If you don’t pass them, Python gives an error.

## **4. Variable-Length Arguments (*args and **kwargs)**
Sometimes, you want a function to accept any number of arguments. Python allows this in two ways:

### **a)  (*args) (Arbitrary Positional Arguments)**
Collects extra positional arguments as a tuple.

Used when you don’t know how many arguments will be passed.

### **b) kwargs (Arbitrary Keyword Arguments)**
Collects extra keyword arguments as a dictionary.

Useful when you want to pass named arguments dynamically.

# **4. Factorial Calculator**
 * Goal: Compute factorial using recursion or loops.
 * *important point:* function is exite on return or print

```python
# Find factorial using Loop
def factorial(num):
  fact = 1
  for i in range(1 ,num+1):
    fact = fact * i
  return (fact)
print(factorial(7))
```

```python
# Find a factorial using Recursion
def factorial2(num):
  if num==0 or num ==1:
    return 1
  else:
     return(num * factorial2(num - 1))

# Calling a Function
print(factorial2(5))
```

# **5. Password Strength Checker**
 * Goal: A function that checks password strength (length, special chars).

```python
# Password Checker
def password_checker(password):
# Check length of password is at least 8 characters
  if len(password) < 8:
    return("The password is too Short")
  elif not any (char.isdigit() for char in password): # Check it has any number
    return ("Password does not have any numbers!")
  elif not any (char.isupper() for char in password): # is have any upper case or capital lettter
    return ("No have any upper case letter!")
  elif not any (char.isalnum() for char in password): # Check it has any special character
    return ("Not have any symbol or special chracter like (! , @ ,#, $)")
  else:
    return ("Strong!") # Retun Strong password

# Calling Function
print(password_checker("1223qweE"))
```

## 2.Farnhit to celcius
* round() only works on numbers, not strings.

* First compute → then round → then format as a string.

---



```python
# Temperature converter from farhit to celcius
def farnhit_to_celcius(f):
  celcius = (f - 32)/1.8 # is act as integer
  celcius =round(celcius)
  celcius =(f"{f}F into celcius is: {celcius}C")
  return (celcius)
# Calling a function
farnhit_to_celcius(22)
```

```python
# Alternative avoiding use of round()
def fahrenheit_to_celsius(f):
    celsius = (f - 32) / 1.8
    return f"{f}F into Celsius is: {celsius:.1f}C"  # Rounds to 1 decimal place

print(fahrenheit_to_celsius(22))  # Output: 22F into Celsius is: -5.6C
```

# **7.List Duplicate Remover**
* Goal: A function that removes duplicates from a list.



## a) Remove duplicates
  * using set method


```python
# Remove duplicates from a list by using set method
def convert_lis(*a):
  remove_duplicates = (set(a)) # convert into set
  return (remove_duplicates)
# Calling function
convert_lis(1,2,3,4,5,5,4,4,7)
```

## b) Remove duplicates
  *  using loop
  * It is slower than other methods

```python
# remove duplicates from a list using loop method
def remove_dupli(*l):
  unique_list=[]
  for i in l:
    if i not in unique_list:
      unique_list.append(i)
  return unique_list
# Calling a function
print(remove_dupli(1,2,3,3,4,5))
```

# **8. Counter**


## a) Vowel Counter
 * Goal: Count vowels in a string.

```python
# Vowels count function in a string
def vowel_counter(v): # Take a string
  count=0  # define count
  for letters in  v: # Iterate each letter
    if letters in "aeiouAEIOU":
      count+=1 # count vowels
  return f"{count} vowels exist."
# Calling Function
vowel_counter("I am very happy and python is easy to learn.")
```

## b) Consonants counter  
  * Goal: Count consonants in a string.



```python
# Consonants counts in a string
def consonants_counter(a): # Define a function
  remove_space=a.replace(" ","") # Remove white spaces from a string
  count=0
  for i in remove_space: # Iterate through each
    if i not in "aeiouAEIOU": # It takes only consonants
      count+=1
  return(count)
# Calling a function
consonants_counter("I am very happy")
```
